```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.datasets import make_blobs

X, y = make_blobs(
    n_samples=600, centers=5, cluster_std=1.2, random_state=0
)

df = pd.DataFrame(X, columns=["income", "spending_score"])

plt.figure(figsize=(8, 6))
plt.scatter(
    df["income"],
    df["spending_score"],
    alpha=0.7,
    edgecolors="k",
    c="tab:blue",
)
plt.title("Розподіл даних клієнтів (income vs spending_score)")
plt.xlabel("Income")
plt.ylabel("Spending Score")
plt.grid(True, linestyle="--", alpha=0.5)
plt.show()

print(df.head())
```
