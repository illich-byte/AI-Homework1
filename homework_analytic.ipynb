```python
import pandas as pd
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

wine = load_wine()

df = pd.DataFrame(data=wine.data, columns=wine.feature_names)
df["target"] = wine.target

X = df.drop(columns=["target"])
y = df["target"]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42, stratify=y
)

print(f"Форма навчальної вибірки: {X_train.shape}")
print(f"Форма тестової вибірки: {X_test.shape}")
print("\nПерші 3 рядки масштабованих ознак у DataFrame:")
print(pd.DataFrame(X_scaled, columns=wine.feature_names).head(3))
```
