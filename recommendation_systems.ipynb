```python
import pandas as pd

books_url = "https://raw.githubusercontent.com/zygmuntz/goodbooks-10k/master/books.csv"
ratings_url = "https://raw.githubusercontent.com/zygmuntz/goodbooks-10k/master/ratings.csv"

print("Завантаження датасетів...")
books_df = pd.read_csv(books_url)
ratings_df = pd.read_csv(ratings_url)

books_df = books_df.drop_duplicates().dropna(subset=["book_id", "title", "authors"])
ratings_df = ratings_df.drop_duplicates().dropna(subset=["book_id", "user_id", "rating"])

books_df = books_df[["book_id", "title", "authors"]]
ratings_df = ratings_df[["book_id", "user_id", "rating"]]

merged_df = pd.merge(ratings_df, books_df, on="book_id", how="inner")

merged_df = merged_df.drop_duplicates().dropna()

print("\nДані успішно підготовлені та об'єднані!")
print(f"Загальна кількість рядків: {len(merged_df):,}")
print("\nПерші 5 рядків об'єднаного датасету:")
print(merged_df.head())
```
