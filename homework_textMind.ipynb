```python
import re
import nltk
import pandas as pd
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download("punkt")
nltk.download("stopwords")

reviews = [
    "Чудовий продукт! Дійсно насолоджувався користуванням щодня на 100%.",
    "Служба підтримки була некорисною та повільною...",
    "Пристойна якість за таку ціну, але доставка зайняла 2 тижні!",
    "Абсолютно дивовижно! Найкраща покупка, яку я зробив у 2024 році.",
    "Жахливий досвід, товар прийшов пошкодженим і зламаним.",
    "Додаток постійно вилітає після останнього оновлення v2.0.",
    "Дуже швидка доставка та чудова упаковка! Настійно рекомендую.",
    "Нормально, нічого особливого. Хоча могло б бути й дешевше.",
    "П'ять зірок! Простий у використанні та супер інтуїтивний інтерфейс.",
    "Не вартує своїх грошей, якість дуже погана...",
    "Надзвичайно задоволений сервісом та швидким часом відповіді!",
    "Продукт працює, як описано, поки що жодних проблем.",
]

stop_words = set(stopwords.words("english"))
tokenized_sentences = []

for text in reviews:
    clean_text = text.lower()
    clean_text = re.sub(r"[^a-z\s]", "", clean_text)

    tokens = word_tokenize(clean_text)

    filtered_tokens = [
        token for token in tokens if token not in stop_words and len(token) > 1
    ]

    if filtered_tokens:
        tokenized_sentences.append(filtered_tokens)

print(
    f"Оброблено речень: {len(tokenized_sentences)}\n"
) 
print("Приклад перших 3 токенізованих речень:")
for i, sentence in enumerate(tokenized_sentences[:3], 1):
    print(f"Речення {i}: {sentence}")
```
