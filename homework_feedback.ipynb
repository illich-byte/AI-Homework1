```python
import re
import pandas as pd

data = {
    "review_id": list(range(1, 13)),
    "text": [
        "Great product! Really enjoyed using it every day 100%.",
        "Customer support was unhelpful and slow...",
        "Decent quality for the price, but shipping took 2 weeks!",
        "Absolutely amazing! Best purchase I've made in 2024.",
        "Terrible experience, the item arrived damaged & broken.",
        "The app crashes constantly after the latest update v2.0.",
        "Very fast delivery and awesome packaging! Highly recommend.",
        "It's okay, nothing special. Could be cheaper though.",
        "Five stars! Easy to use and super intuitive user interface.",
        "Not worth the money, quality is very poor...",
        "Extremely satisfied with the service and fast response time!",
        "The product works as described, no issues so far.",
    ],
}

df_initial = pd.DataFrame(data)
df_initial.to_csv("reviews.csv", index=False)

df = pd.read_csv("reviews.csv")

def clean_text(text):
    text = text.lower()  
    text = re.sub(
        r"[^a-z\s]", "", text
    )  
    text = re.sub(r"\s+", " ", text).strip()  
    return text


df["clean_text"] = df["text"].apply(clean_text)

total_rows = len(df)
df["word_count"] = df["clean_text"].apply(lambda x: len(x.split()))
avg_word_count = df["word_count"].mean()

print(f"Кількість рядків: {total_rows}")
print(f"Середня довжина відгуків (у словах): {avg_word_count:.2f}\n")
print("Результат обробки даних:")
print(df[["text", "clean_text", "word_count"]].head())
```
