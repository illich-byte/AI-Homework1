```python
import pandas as pd

from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
housing = fetch_california_housing(as_frame=True)

df = housing.frame

X = df[["MedInc"]]
y = df["MedHouseVal"]

print(df.head())
```

```python
poly = PolynomialFeatures(degree=2)

X_poly = poly.fit_transform(X)

print("Розмірність після перетворення:", X_poly.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_poly,
    y,
    test_size=0.2,
    random_state=42
)
```

```python
model = LinearRegression()

model.fit(X_train, y_train)

y_pred = model.predict(X_test)
```

```python
r2 = r2_score(y_test, y_pred)

print("R² =", round(r2, 4))
```

```python
result = pd.DataFrame({
    "Модель": ["Polynomial Regression"],
    "Ознака": ["MedInc"],
    "Ступінь": [2],
    "R²": [round(r2, 4)]
})

print(result)
```

- Завантажено датасет California Housing.
- Використано лише ознаку **MedInc**.
- Створено поліноміальні ознаки другого степеня.
- Дані розділено у співвідношенні **80/20**.
- Навчено модель поліноміальної регресії.
- Обчислено коефіцієнт детермінації **R²**.
